```python
import time

def q():
    print("답이 틀려")
    time.sleep(3)

def a():
    print("하지만 빨랐죠")

q()
a()
```

```python
import asyncio
async def q():
    print("답이 틀려")
    await asyncio.sleep(3)

async def a():
    print("하지만 빨랐죠")

async def main():
    await asyncio.gather(q(),a())
```

```python
asyncio.run(main())
```
